# RESULTADOS

Modelo final  -> Random forest optimizado

Precision (churn): 0.95
Recall (churn): 0.02
F1-score (churn): 0.03
ROC-AUC: 0.806

El modelo Random Forest optimizado obtuvo los mejores resultados globales entre todos los algoritmos evaluados. La métrica ROC-AUC alcanzó un valor de 0.806, indicando una buena capacidad de discriminación entre clientes que abandonan el servicio y clientes que permanecen en la compañía.

Sin embargo, debido al fuerte desbalanceo de la variable objetivo, el modelo mostró dificultades para identificar correctamente los casos de churn utilizando el umbral de clasificación estándar de 0.5. Aunque la precisión para la clase positiva fue elevada, el recall resultó reducido, lo que indica que una parte importante de los clientes que finalmente abandonaron la compañía no fueron detectados.

Se evaluaron distintos algoritmos de clasificación con el objetivo de identificar la solución con mayor capacidad predictiva. Entre los modelos analizados se incluyeron una regresión logística, Random Forest, Gradient Boosting y XGBoost.

Los resultados muestran que el Random Forest optimizado mediante Grid Search obtuvo el mejor rendimiento global, alcanzando un valor ROC-AUC de 0.806. Este resultado indica que la combinación de múltiples árboles de decisión permitió capturar relaciones no lineales presentes en los datos y mejorar la capacidad predictiva respecto a los modelos alternativos.

El análisis de importancia de variables permitió identificar los principales factores asociados al riesgo de abandono. Entre las variables con mayor relevancia destacaron la antigüedad del cliente, los ingresos estimados, el valor económico del cliente, el ratio gasto-ingreso y diversas métricas relacionadas con la calidad de red.

También resultaron relevantes las variables derivadas creadas durante el proceso de feature engineering, como problemas_red, estres_red, valor_cliente y delta_calidad, lo que confirma que la combinación de múltiples fuentes de información aporta valor predictivo adicional respecto al uso exclusivo de variables originales.


Con el objetivo de evaluar la capacidad de generalización del modelo, se realizaron validaciones adicionales mediante validación cruzada y particiones temporales.

La validación cruzada obtuvo un ROC-AUC medio de 0.646, mientras que la validación temporal utilizando datos históricos para predecir periodos futuros alcanzó un ROC-AUC de 0.656.

Estos resultados son inferiores al valor obtenido mediante la partición aleatoria entrenamiento-prueba. Este comportamiento sugiere que parte del rendimiento observado inicialmente puede estar asociado a la alta similitud existente entre observaciones de un mismo cliente en distintos meses. No obstante, los resultados obtenidos en los escenarios más exigentes siguen mostrando capacidad predictiva superior al azar y permiten identificar patrones relevantes asociados al churn.

Debido al fuerte desbalanceo de la variable objetivo, se analizó el impacto de distintos umbrales de clasificación sobre la capacidad de detección de clientes con riesgo de abandono.

Los resultados muestran que la reducción del umbral incrementa significativamente el recall de la clase churn, permitiendo identificar una mayor proporción de clientes que finalmente abandonan la compañía. En particular, un umbral de 0.15 permitió detectar aproximadamente el 25% de los casos de churn, aunque a costa de un incremento en el número de falsos positivos.

Desde una perspectiva de negocio, esta estrategia puede resultar adecuada en contextos donde el coste de una acción preventiva es inferior al coste asociado a la pérdida de un cliente.
